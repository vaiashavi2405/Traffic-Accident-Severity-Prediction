```python
# ============================================================
# TRAFFIC ACCIDENT SEVERITY PREDICTION
# Machine Learning Project
# Dataset: Road Accident Severity in India
# Target: Accident_severity
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
import os
import joblib

warnings.filterwarnings("ignore")

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)


# ============================================================
# 2. DATASET PATH
# ============================================================

# If Road.csv is inside dataset folder and notebook is inside notebook folder:
DATA_PATH = ("Road.csv")

# If Road.csv is in the same folder as notebook, use:
# DATA_PATH = "Road.csv"


# ============================================================
# 3. CREATE MODELS FOLDER
# ============================================================

os.makedirs("../models", exist_ok=True)

# If notebook and dataset are in same folder, use:
# os.makedirs("models", exist_ok=True)


# ============================================================
# 4. LOAD DATASET
# ============================================================

print("=" * 70)
print("LOADING DATASET")
print("=" * 70)

try:
    df = pd.read_csv(DATA_PATH)
except FileNotFoundError:
    print("\n❌ Dataset not found!")
    print("Current path:", DATA_PATH)
    print("\nCheck your folder structure.")
    raise


print("\nDataset loaded successfully!")

print("\nShape of dataset:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# 5. BASIC INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# ============================================================
# 6. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing = df.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

if len(missing) > 0:
    print(missing)
else:
    print("No missing values found.")


# ============================================================
# 7. CHECK DUPLICATES
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE ROWS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# Remove duplicates
if duplicates > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")

print("New shape:", df.shape)


# ============================================================
# 8. TARGET COLUMN
# ============================================================

TARGET = "Accident_severity"

print("\n" + "=" * 70)
print("TARGET VARIABLE")
print("=" * 70)

print("Target column:", TARGET)

print("\nTarget values:")
print(df[TARGET].value_counts())


# ============================================================
# 9. TARGET DISTRIBUTION
# ============================================================

plt.figure(figsize=(8, 5))

df[TARGET].value_counts().plot(
    kind="bar"
)

plt.title("Accident Severity Distribution")
plt.xlabel("Accident Severity")
plt.ylabel("Number of Accidents")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# Percentage distribution

target_percentage = (
    df[TARGET]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nTarget percentage distribution:")
print(target_percentage)


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(df.describe(include="all").T)


# ============================================================
# 11. FEATURE ENGINEERING - TIME
# ============================================================

print("\n" + "=" * 70)
print("TIME FEATURE ENGINEERING")
print("=" * 70)

if "Time" in df.columns:

    # Convert Time into datetime
    time_parsed = pd.to_datetime(
        df["Time"],
        format="%H:%M:%S",
        errors="coerce"
    )

    # Extract hour
    df["Hour"] = time_parsed.dt.hour

    # Create time period
    def get_time_period(hour):

        if pd.isna(hour):
            return "Unknown"

        hour = int(hour)

        if 5 <= hour < 12:
            return "Morning"

        elif 12 <= hour < 17:
            return "Afternoon"

        elif 17 <= hour < 21:
            return "Evening"

        else:
            return "Night"

    df["Time_period"] = df["Hour"].apply(get_time_period)

    # Drop original Time column
    df.drop(columns=["Time"], inplace=True)

    print("Time feature engineering completed.")

else:
    print("Time column not found.")


# ============================================================
# 12. CHECK HOUR DISTRIBUTION
# ============================================================

if "Hour" in df.columns:

    plt.figure(figsize=(10, 5))

    df["Hour"].value_counts().sort_index().plot(
        kind="bar"
    )

    plt.title("Accidents by Hour")
    plt.xlabel("Hour of Day")
    plt.ylabel("Number of Accidents")
    plt.tight_layout()
    plt.show()


# ============================================================
# 13. TIME PERIOD DISTRIBUTION
# ============================================================

if "Time_period" in df.columns:

    plt.figure(figsize=(8, 5))

    df["Time_period"].value_counts().plot(
        kind="bar"
    )

    plt.title("Accidents by Time Period")
    plt.xlabel("Time Period")
    plt.ylabel("Number of Accidents")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# 14. IMPORTANT CATEGORICAL FEATURES
# ============================================================

important_categorical_features = [
    "Day_of_week",
    "Age_band_of_driver",
    "Sex_of_driver",
    "Driving_experience",
    "Road_surface_conditions",
    "Light_conditions",
    "Weather_conditions",
    "Type_of_collision",
    "Cause_of_accident"
]

for col in important_categorical_features:

    if col in df.columns:

        plt.figure(figsize=(10, 5))

        df[col].value_counts().head(10).plot(
            kind="bar"
        )

        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Count")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()


# ============================================================
# 15. NUMERIC FEATURES
# ============================================================

numeric_candidates = [
    "Number_of_vehicles_involved",
    "Number_of_casualties",
    "Hour"
]

for col in numeric_candidates:

    if col in df.columns:

        plt.figure(figsize=(8, 5))

        plt.hist(
            df[col].dropna(),
            bins=20
        )

        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()


# ============================================================
# 16. ACCIDENT SEVERITY VS WEATHER
# ============================================================

if "Weather_conditions" in df.columns:

    cross_weather = pd.crosstab(
        df["Weather_conditions"],
        df[TARGET]
    )

    cross_weather.plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Accident Severity vs Weather Conditions")
    plt.xlabel("Weather Conditions")
    plt.ylabel("Number of Accidents")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# 17. ACCIDENT SEVERITY VS LIGHT CONDITIONS
# ============================================================

if "Light_conditions" in df.columns:

    cross_light = pd.crosstab(
        df["Light_conditions"],
        df[TARGET]
    )

    cross_light.plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Accident Severity vs Light Conditions")
    plt.xlabel("Light Conditions")
    plt.ylabel("Number of Accidents")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# 18. ACCIDENT SEVERITY VS DRIVING EXPERIENCE
# ============================================================

if "Driving_experience" in df.columns:

    cross_driving = pd.crosstab(
        df["Driving_experience"],
        df[TARGET]
    )

    cross_driving.plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Accident Severity vs Driving Experience")
    plt.xlabel("Driving Experience")
    plt.ylabel("Number of Accidents")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# 19. REMOVE POSSIBLE ID COLUMNS
# ============================================================

possible_id_columns = [
    "ID",
    "id",
    "Accident_ID",
    "accident_id"
]

columns_to_drop = [
    col for col in possible_id_columns
    if col in df.columns
]

if columns_to_drop:

    df.drop(
        columns=columns_to_drop,
        inplace=True
    )

    print("\nRemoved ID columns:", columns_to_drop)


# ============================================================
# 20. SEPARATE FEATURES AND TARGET
# ============================================================

X = df.drop(
    columns=[TARGET]
)

y = df[TARGET]


print("\n" + "=" * 70)
print("FEATURE / TARGET SPLIT")
print("=" * 70)

print("X shape:", X.shape)
print("y shape:", y.shape)


# ============================================================
# 21. IDENTIFY NUMERIC AND CATEGORICAL FEATURES
# ============================================================

numeric_features = X.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()


print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 22. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


print("\n" + "=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# ============================================================
# 23. PREPROCESSING PIPELINE
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)


# Handle sklearn version difference
try:

    categorical_encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:

    categorical_encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )


categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            categorical_encoder
        )
    ]
)


preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 24. DEFINE MACHINE LEARNING MODELS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=12,
        min_samples_split=10,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_split=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}


# ============================================================
# 25. TRAIN MODELS
# ============================================================

results = []

trained_pipelines = {}

print("\n" + "=" * 70)
print("MODEL TRAINING")
print("=" * 70)


for model_name, model in models.items():

    print("\nTraining:", model_name)

    pipeline = Pipeline(
        steps=[
            (
                "preprocessor",
                preprocessor
            ),
            (
                "model",
                model
            )
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    trained_pipelines[model_name] = pipeline

    y_pred = pipeline.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average="weighted",
        zero_division=0
    )

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })

    print(
        f"Accuracy  : {accuracy:.4f}"
    )

    print(
        f"Precision : {precision:.4f}"
    )

    print(
        f"Recall    : {recall:.4f}"
    )

    print(
        f"F1 Score  : {f1:.4f}"
    )


# ============================================================
# 26. MODEL COMPARISON
# ============================================================

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)


print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(results_df)


# ============================================================
# 27. MODEL COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

x = np.arange(len(results_df))
width = 0.2

plt.bar(
    x - 0.3,
    results_df["Accuracy"],
    width,
    label="Accuracy"
)

plt.bar(
    x - 0.1,
    results_df["Precision"],
    width,
    label="Precision"
)

plt.bar(
    x + 0.1,
    results_df["Recall"],
    width,
    label="Recall"
)

plt.bar(
    x + 0.3,
    results_df["F1 Score"],
    width,
    label="F1 Score"
)

plt.xticks(
    x,
    results_df["Model"],
    rotation=20
)

plt.ylim(0, 1.05)

plt.title("Machine Learning Model Comparison")
plt.ylabel("Score")
plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 28. SELECT BEST MODEL
# ============================================================

best_model_name = results_df.iloc[0]["Model"]

best_pipeline = trained_pipelines[
    best_model_name
]


print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print("Best Model:", best_model_name)
print(
    "Best F1 Score:",
    round(
        results_df.iloc[0]["F1 Score"],
        4
    )
)


# ============================================================
# 29. DETAILED EVALUATION
# ============================================================

y_pred_best = best_pipeline.predict(
    X_test
)


print("\n" + "=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        y_pred_best,
        zero_division=0
    )
)


# ============================================================
# 30. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred_best,
    labels=best_pipeline.classes_
)


print("\nConfusion Matrix:")
print(cm)


plt.figure(figsize=(8, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=best_pipeline.classes_
)

disp.plot(
    xticks_rotation=30
)

plt.title(
    f"Confusion Matrix - {best_model_name}"
)

plt.tight_layout()
plt.show()


# ============================================================
# 31. FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE")
print("=" * 70)


best_estimator = best_pipeline.named_steps["model"]

if hasattr(best_estimator, "feature_importances_"):

    # Get transformed feature names
    fitted_preprocessor = (
        best_pipeline.named_steps["preprocessor"]
    )

    feature_names = (
        fitted_preprocessor
        .get_feature_names_out()
    )

    importances = (
        best_estimator
        .feature_importances_
    )

    feature_importance_df = pd.DataFrame({

        "Feature": feature_names,

        "Importance": importances

    })

    feature_importance_df = (
        feature_importance_df
        .sort_values(
            by="Importance",
            ascending=False
        )
        .reset_index(drop=True)
    )


    print("\nTop 20 Important Features:")

    display(
        feature_importance_df.head(20)
    )


    # Plot
    top_features = (
        feature_importance_df
        .head(20)
        .sort_values(
            by="Importance"
        )
    )

    plt.figure(figsize=(10, 8))

    plt.barh(
        top_features["Feature"],
        top_features["Importance"]
    )

    plt.title(
        f"Top 20 Feature Importances - {best_model_name}"
    )

    plt.xlabel("Importance")

    plt.tight_layout()

    plt.show()

else:

    print(
        "Feature importance is not available for",
        best_model_name
    )


# ============================================================
# 32. SAVE BEST MODEL
# ============================================================

MODEL_PATH = "../models/traffic_accident_severity_model.pkl"

joblib.dump(
    best_pipeline,
    MODEL_PATH
)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(
    "Model saved successfully at:",
    MODEL_PATH
)


# ============================================================
# 33. SAVE MODEL COMPARISON
# ============================================================

RESULTS_PATH = "../models/model_comparison.csv"

results_df.to_csv(
    RESULTS_PATH,
    index=False
)

print(
    "Model comparison saved at:",
    RESULTS_PATH
)


# ============================================================
# 34. SAVE TARGET CLASSES
# ============================================================

TARGET_MAPPING_PATH = "../models/target_classes.pkl"

joblib.dump(
    list(best_pipeline.classes_),
    TARGET_MAPPING_PATH
)

print(
    "Target classes saved at:",
    TARGET_MAPPING_PATH
)


# ============================================================
# 35. CHECK SAVED FILES
# ============================================================

print("\n" + "=" * 70)
print("SAVED FILES")
print("=" * 70)

print(
    "\nModel:",
    MODEL_PATH
)

print(
    "Comparison:",
    RESULTS_PATH
)

print(
    "Target classes:",
    TARGET_MAPPING_PATH
)


# ============================================================
# 36. SAMPLE PREDICTION
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE PREDICTION")
print("=" * 70)


# Take one real test sample
sample = X_test.iloc[[0]]

actual_value = y_test.iloc[0]

prediction = best_pipeline.predict(
    sample
)[0]


print("\nActual Accident Severity:")
print(actual_value)

print("\nPredicted Accident Severity:")
print(prediction)


# ============================================================
# 37. PREDICTION PROBABILITY
# ============================================================

if hasattr(
    best_pipeline,
    "predict_proba"
):

    probabilities = (
        best_pipeline
        .predict_proba(sample)[0]
    )

    probability_df = pd.DataFrame({

        "Accident Severity":
            best_pipeline.classes_,

        "Probability":
            probabilities

    })

    probability_df["Probability"] = (
        probability_df["Probability"] * 100
    ).round(2)


    print("\nPrediction probabilities:")

    display(
        probability_df
    )


# ============================================================
# 38. LOAD MODEL AGAIN - TEST SAVED MODEL
# ============================================================

print("\n" + "=" * 70)
print("TESTING SAVED MODEL")
print("=" * 70)


loaded_model = joblib.load(
    MODEL_PATH
)


loaded_prediction = (
    loaded_model.predict(sample)[0]
)


print(
    "Prediction using loaded model:",
    loaded_prediction
)


# ============================================================
# 39. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("\nProject Name:")
print("Traffic Accident Severity Prediction")

print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Features:")
print(X.shape[1])

print("\nBest Model:")
print(best_model_name)

print(
    "\nBest F1 Score:",
    round(
        results_df.iloc[0]["F1 Score"],
        4
    )
)

print("\nTarget Classes:")
print(
    list(best_pipeline.classes_)
)

print("\nGenerated Files:")
print("1. traffic_accident_severity_model.pkl")
print("2. model_comparison.csv")
print("3. target_classes.pkl")

print("\n" + "=" * 70)
print("DONE!")
print("=" * 70)
```
